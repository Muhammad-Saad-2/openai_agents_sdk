```python
#direct call to the LLM

import requests
import json

response = requests.post(
    url = f"https://openrouter.ai/api/v1/chat/completions",
    headers = {
        "Authorization": f"Bearer {api_key}"
    },
    data = json.dumps({
        "model": model,
        "messages": [
            {
                "role": "user",
                "content": "What is the meaning of life?"
            }
        ]
    })

)
print(response.json())
```

```python
#using openrouter with OpenAi agents sdk

from agents import Agent, Runner, OpenAIChatCompletionsModel, AsyncOpenAI, set_tracing_disabled
from agents.run import RunConfig
import asyncio

set_tracing_disabled(disabled=True)

external_client = AsyncOpenAI(
    api_key = api_key,
    base_url = "https://openrouter.ai/api/v1",
  )

model = OpenAIChatCompletionsModel(
    model = "deepseek/deepseek-r1-distill-qwen-7b",
    openai_client= external_client,
)

config = RunConfig(
    model = model,
    model_provider = external_client
)

biology_agent = Agent(
    name = "biology agent",
    instructions= "you're a biology professor highly skilled in your domain having 30 years of proven experience globally known for your services and contribution in the field of biology",
    model = model,
)

async def main():
  output = await Runner.run(biology_agent, "explain the components of brain", run_config=config)
  print(output.final_output)

if __name__ == "__main__":
  asyncio.run(main())


```
