```python
def print_details(**kwargs):
    """
    This function accepts any number of named (keyword) arguments
    and prints them.
    """
    print("Here are the details you provided:")
    for key, value in kwargs.items():
        print(f"- {key}: {value}")

# --- Using the function ---

print("Example 1: Providing name and age")
print_details(name="Alice", age=30)
print("-" * 30)

print("Example 2: Providing city and temperature")
print_details(city="New York", temperature="25C")
print("-" * 30)

print("Example 3: Providing various random details")
print_details(product="Laptop", price=1200, color="Silver", brand="TechCorp")
print("-" * 30)

print("Example 4: No extra details provided")
print_details()
print("-" * 30)
```
